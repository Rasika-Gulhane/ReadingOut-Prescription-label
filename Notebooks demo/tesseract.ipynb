````python

def perform_ocr(image_path):
    try:
        # Open the image file
        with Image.open(image_path) as img:
            # Convert the PIL image to an OpenCV image
            opencv_image = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

            # Normalize the image (you can customize these normalization steps)
            normalized_image = cv2.normalize(opencv_image, None, alpha=25, beta=225, norm_type=cv2.NORM_MINMAX)

            # Convert the normalized image back to PIL format
            normalized_pil_image = Image.fromarray(cv2.cvtColor(normalized_image, cv2.COLOR_BGR2RGB))

            # Perform OCR using pytesseract on the normalized image
            text = pytesseract.image_to_string(opencv_image, config='--psm 6')

            # Print the extracted text
            print("OCR Output:")
            print(text)

            # You can further process or return the text as needed
            return text

    except Exception as e:
        print(f"Error during OCR: {str(e)}")
````

### Improved Version

````python

def improve_resolution(image_path, scale_factor):
    # Read the image
    img = cv2.imread(image_path)

    # Resize the image using interpolation
    new_width = int(img.shape[1] * scale_factor)
    new_height = int(img.shape[0] * scale_factor)
    resized_img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_LINEAR)

    # Save or display the improved resolution image
    cv2.imwrite("improved_resolution.jpg", resized_img)

improve_resolution(image_path, 0.5)
````

````python
import os
from google.cloud import vision

# Set the environment variable
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "peak-castle-416420-afc2d99ff254.json"

def detect_document(path):
    """Detects document features in an image."""
    # Use the service account file for authentication
    client = vision.ImageAnnotatorClient.from_service_account_file(os.environ["GOOGLE_APPLICATION_CREDENTIALS"])

    with open(path, "rb") as image_file:
        content = image_file.read()
    image = vision.Image(content=content)
    response = client.document_text_detection(image=image)

    for page in response.full_text_annotation.pages:
        for block in page.blocks:
            print(f"\nBlock confidence: {block.confidence}\n")
            for paragraph in block.paragraphs:
                print("Paragraph confidence: {}".format(paragraph.confidence))
                for word in paragraph.words:
                    word_text = "".join([symbol.text for symbol in word.symbols])
                    print(
                        "Word text: {} (confidence: {})".format(
                            word_text, word.confidence
                        )
                    )
                    for symbol in word.symbols:
                        print(
                            "\tSymbol: {} (confidence: {})".format(
                                symbol.text, symbol.confidence
                            )
                        )
    if response.error.message:
        raise Exception(
            "{}\nFor more info on error messages, check: "
            "https://cloud.google.com/apis/design/errors".format(response.error.message)
        )
````

````python
from langchain.chat_models import ChatOpenAI
from langchain.llms import OpenAI
import openai
import os
from kor.extraction import create_extraction_chain
from kor.nodes import Object, Text, Number
from langchain_openai import ChatOpenAI
````

````python



# Set your OpenAI GPT-3 API key

mykey = os.environ.get('OPEN_AI_API_KEY')   #set key set to environment variable

from openai import OpenAI
client = OpenAI(api_key=mykey)
````

````python
llm = ChatOpenAI(
#     model_name="gpt-3.5-turbo", # Cheaper but less reliable
    model_name="gpt-4",
    temperature=0,
    max_tokens=2000,
    openai_api_key = mykey
)
````

````python
Medical_schema = Object(
    # This what will appear in your output. It's what the fields below will be nested under.
    # It should be the parent of the fields below. Usually it's singular (not plural)
    id="Medicine_name",
    
    # Natural language description about your object
    description="Medicines information from prescription",
    
    # Fields you'd like to capture from a piece of text about your object.
    attributes=[
        Text(
            id="Medicine_name",
            description="Name of prescribed medicine.",
        )
    ],
    
    # Examples help go a long way with telling the LLM what you need
    examples=[
        ("Alice and Bob are friends", [{"first_name": "Alice"}, {"first_name": "Bob"}])
    ]
)
````

````python


medicine_schema = Object(
    id="medicine",
    description="Information about a prescribed medicine",
    attributes=[
        Text(
            id="medicine_name",
            description="The name of the prescribed medicine.",
        ),
        Text(
            id="content_value",
            description="The dosage content value of the medicine (e.g., 100mg).",
        ),
     
    ],
    examples=[
        (
            "Betaloc 100mg - 1 tab BID",
            [{"name": "Betaloc", "content_value": "100mg"}]
        ),
        (
            "Dorzolamidum 10 - 1 tab BID",
            [{"name": "Dorzolamidum", "content_value": "10mg"}]
        ),
        # Add more examples as needed
    ]
)
````

````python
chain = create_extraction_chain(llm, medicine_schema )
````
