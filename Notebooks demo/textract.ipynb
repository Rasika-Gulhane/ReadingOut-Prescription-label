```python
import cv2

def improve_resolution(image_path, scale_factor):
    # Read the image
    img = cv2.imread(image_path)

    # Resize the image using interpolation
    new_width = int(img.shape[1] * scale_factor)
    new_height = int(img.shape[0] * scale_factor)
    resized_img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_LINEAR)

    # Save or display the improved resolution image
    cv2.imwrite("improved_resolution_textract.jpg", resized_img)

improve_resolution(img, 1)
```

```python
import textract



try:
    # Open the image file in binary mode
    with open(improved_img, "rb") as img_file:
        # Attempt text extraction using textract
        extraction_result = textract.process(img_file, method='tesseract', encoding='utf-8', raw=True)

        # Check if extraction was successful and confidence level is available
        if extraction_result and 'metadata' in extraction_result:
            metadata = extraction_result['metadata']
            if 'Confidence' in metadata:
                confidence = metadata['Confidence']
                print(f"Text extraction successful with confidence level: {confidence}%")
            else:
                print("Text extraction successful, but confidence level not available.")
        else:
            print("Text extraction unsuccessful or metadata not available.")

        # Decode and print the extracted text
        extracted_text = extraction_result.decode('utf-8')
        print(extracted_text)
except Exception as e:
    # Handle potential errors during extraction
    print(f"An error occurred during text extraction: {e}")

```

```python
import boto3

# # Replace with your AWS credentials
# access_key_id = "your_access_key_id"
# secret_access_key = "your_secret_access_key"

# Create Textract client
client = boto3.client(
    "textract",
    # aws_access_key_id=access_key_id,
    # aws_secret_access_key=secret_access_key,
)

# Function to extract text from image using Textract
def extract_text_from_image(image_path):
    with open(image_path, "rb") as image_file:
        image_bytes = image_file.read()

    response = client.analyze_document(Document={"Bytes": image_bytes})
    blocks = response["Blocks"]

    extracted_text = ""
    for block in blocks:
        if block["BlockType"] == "LINE":
            extracted_text += block["Text"] + "\n"

    return extracted_text

# Example usage
image_path = "path/to/your/image.jpg"
extracted_text = extract_text_from_image(improved_img)

if extracted_text:
    print("Extracted text:")
    print(extracted_text)
else:
    print("No text found in the image.")
```
